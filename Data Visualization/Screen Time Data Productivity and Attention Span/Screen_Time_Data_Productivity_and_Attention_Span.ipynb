# Screen Time, Productivity and Attention Span

## Introduction

This analysis explores the screen time dataset and its relationship
with productivity and attention span. The goal is to answer the following questions:

- Is there a relationship between screen time and attention span?
- Does the type of app used influence productivity?
- How does notification handling affect attention and productivity?
- Are there differences between students and professionals?

**Tools used:** Python · Pandas · Plotly  
**Dataset:** [Screen Time Data — Kaggle](https://www.kaggle.com/datasets/mexwell/screen-time-and-productivity)

The dataset "Screen Time Data: Productivity and Attention Span" (https://www.kaggle.com/datasets/muhammadalirazazaidi/screen-time-data-productivity-and-attention-span) contains 200 records representing participants from different age groups, occupations and usage patterns. Each record includes information about daily screen time, type of apps used, notification handling strategy, work strategy, and self-reported attention and productivity scores.

## 1. Library Imports

```python
import pandas as pd
import numpy as np
import plotly.express as px
```

## 2. Dataset Loading

The file `data.csv` downloaded from Kaggle is used.
The first rows are displayed below to understand its structure.

```python
# Load the dataset
df = pd.read_csv("data.csv")

# Quick view to verify the data
df.head()
```

## 3. Data Preparation

```python
df.info()
```

```python
# Count null values per variable in percentages
(df.isnull().mean() * 100).round(2)
```

```python
# An unnecessary column called "Unnamed: 0" is found and removed.
df = df.drop(columns=["Unnamed: 0"])

# Some null values are also found, but since there are few records (200 people),
# this represents: Environment = 1% (2 people), Work Strategy = 1% (2 people).
# These rows are dropped.
df = df.dropna()
```

```python
# Count null values per variable in percentages after cleaning
(df.isnull().mean() * 100).round(2)
```

```python
# Check unique values for each column
for col in df.columns:
    print(f"\nColumn: {col}")
    print(df[col].unique())
```

## 4. Variable Transformation

```python
# Average Screen Time: Has values like ['More than 10' '8-10' '4-6' '6-8' '2-4' 'Less than 2']
# Converted to numeric variable based on the average of both values in each range.
screen_map = {
    "Less than 2": 1,
    "2-4": 3,
    "4-6": 5,
    "6-8": 7,
    "8-10": 9,
    "More than 10": 11
}
df["Screen_Time_Hours"] = df["Average Screen Time"].map(screen_map)
```

```python
# Attention Span: Converted to ordinal numeric variable
# Values: ['10-30 minutes' 'More than 1 hour' '30-60 minutes' 'Less than 10 minutes']
attention_map = {
    "Less than 10 minutes": 1,
    "10-30 minutes": 2,
    "30-60 minutes": 3,
    "More than 1 hour": 4
}
df["Attention_Score"] = df["Attention Span"].map(attention_map)
```

```python
# Productivity: Converted to ordinal variable
productivity_map = {
    "Unproductive, i might not have completed the task and got carried away": 1,
    "Moderately productive": 2,
    "Extremely productive, i efficiently complete my tasks": 3
}

df["Productivity_Score"] = df["Productivity"].map(productivity_map)
```

```python
# Age Group: Converted to ordered category
age_order = ["Below 18", "18-24", "25-34", "35-44", "45 and above"]

df["Age Group"] = pd.Categorical(
    df["Age Group"],
    categories=age_order,
    ordered=True
)
```

```python
# Gender: Converted to simple category
df["Gender"] = df["Gender"].astype("category")
```

```python
# Education Level: Converted to ordinal variable
edu_order = ["High school or below", "Undergraduate", "Graduate"]

df["Education Level"] = pd.Categorical(
    df["Education Level"],
    categories=edu_order,
    ordered=True
)
```

```python
# Occupation: Converted to category
df["Occupation"] = df["Occupation"].astype("category")
```

```python
# Device, Screen Activity, App Category, Screen Time Period: All converted to nominal categories
cols_nominal = [
    "Device",
    "Screen Activity",
    "App Category",
    "Screen Time Period",
    "Environment",
    "Work Strategy",
    "Notification Handling"
]

for col in cols_nominal:
    df[col] = df[col].astype("category")
```

```python
# Usage of Productivity Apps: Converted to ordinal
apps_map = {
    "No, i do not use them": 0,
    "Yes, but i did not find them of any help": 1,
    "Yes, they are extremely helpful": 2
}

df["Productivity_App_Score"] = df["Usage of Productivity Apps"].map(apps_map)
```

```python
# Check unique values for each variable again
for col in df.columns:
    print(f"\nColumn: {col}")
    print(df[col].unique())
```

The key variables to work with are:

* Screen_Time_Hours (daily average screen time): 11, 9, 5, 7, 3 or 1
* Attention_Score (attention span): 1 (Less than 10 min), 2 (10-30 min), 3 (30-60 min), 4 (More than 1 hour)
* Productivity_Score: 1 (Unproductive), 2 (Moderately productive), 3 (Extremely productive)

## 5. Charts and Correlations

### 5.1. Screen Time vs Attention Span

```python
df[["Screen_Time_Hours", "Attention_Score"]].corr()
```

Correlation between Screen Time and Attention Span: 0.096.

Conclusion: The correlation is very weak, positive and practically non-existent. There is no significant linear relationship between the hours spent in front of screens and the self-reported attention span.

### 5.2. Screen Time vs Productivity

```python
df[["Screen_Time_Hours", "Productivity_Score"]].corr()
```

Correlation between Screen Time and Productivity: 0.010.

Conclusion: There is also no correlation between screen time and the person's productivity. The data suggests that the amount of time in front of screens does not linearly determine self-perceived productivity.

* 35–44: 7.55 hours

### 5.4. Screen Usage Period vs Attention Span

Conclusion: No significant differences in attention span are observed based on the time of day screens are used.

### 5.5. Productivity vs Notification Handling

```python
df.groupby("Notification Handling")["Productivity_Score"].mean()
```

Productivity according to how the person handles notifications:

Attention span grouped by work strategy:

### 5.7. Barplot: Number of People by Daily Screen Time

```python
import plotly.express as px
import pandas as pd

# Explicit order for categories
age_order = ["Below 18", "18-24", "25-34", "35-44", "45 and above"]
screen_order = ["Less than 2", "2-4", "4-6", "6-8", "8-10", "More than 10"]
period_order = ["Morning (6 AM-12 PM)", "Afternoon (12 PM-6 PM)", "Evening (6 PM-10 PM)", "Night (10 PM-6 AM)"]
```

```python
counts = (
    df["Average Screen Time"]
    .value_counts(normalize=True)
    .reindex(screen_order)
    .reset_index()
)

counts.columns = ["Average Screen Time", "Percentage"]
counts["Percentage"] = counts["Percentage"] * 100

fig1 = px.bar(
    counts,
    x="Average Screen Time",
    y="Percentage",
    category_orders={"Average Screen Time": screen_order},
    text=counts["Percentage"].round(1).astype(str) + "%",
    title="Distribution of Daily Screen Time",
    color_discrete_sequence=["#1f77b4"]
)

fig1.update_layout(
    template="simple_white",
    xaxis_title="Daily Screen Time",
    yaxis_title="Percentage of Participants (%)"
)

fig1.show()
fig1.write_image("barplot_tiempo_pantalla.png")
```

![Barplot screen time](barplot_tiempo_pantalla.png)

Conclusion:
* Most participants report between 4 and 8 hours of daily screen use, reflecting high digital exposure
* Very few participants report less than 2 hours or more than 10 hours per day

## 6. Final Conclusions

- There is no significant correlation between screen time and attention span or productivity
- **Notification handling** is the factor that most impacts productivity: those who ignore them until completing a task show better performance
- **Productivity apps** do not guarantee better results on their own
- **Professionals** spend more hours in front of screens than students, but both groups show similar levels of attention and productivity
- Screen use oriented toward **academic or work tasks** is associated with higher attention than entertainment use
- **Young adults (25–34)** are the group with the highest digital exposure
